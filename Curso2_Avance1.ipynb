```python
class NotebookInput:
    def readline(self, *args):
        try:
            return input() + "\n"
        except EOFError:
            return ""
    def __iter__(self):
        return self
    def __next__(self):
        line = self.readline()
        if not line:
            raise StopIteration
        return line

from Veracity_M1 import main
_ = main(input_stream=NotebookInput())


```
